```python
# ── PHASE 2: Face Detection ──────────────────────────────────────────────────
# Goal: Upload any photo and draw a box around every face found in it

import cv2
import face_recognition
import numpy as np
from google.colab import files
from IPython.display import display
import matplotlib.pyplot as plt

# Step 1: Upload an image from your computer
print("📂 Please upload a photo with one or more faces...")
uploaded = files.upload()

# Step 2: Read the uploaded image
filename = list(uploaded.keys())[0]          # get the filename
image_bgr = cv2.imread(filename)             # OpenCV reads as BGR format
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)  # convert to RGB

print(f"✅ Image loaded: {filename}")
print(f"📐 Image size: {image_rgb.shape[1]}x{image_rgb.shape[0]} pixels")

# Step 3: Detect face LOCATIONS in the image
# face_recognition finds the (top, right, bottom, left) of each face box
face_locations = face_recognition.face_locations(image_rgb)

print(f"\n🔍 Faces detected: {len(face_locations)}")

# Step 4: Draw a rectangle around each detected face
image_display = image_rgb.copy()

for i, (top, right, bottom, left) in enumerate(face_locations):
    # Draw blue rectangle around each face
    cv2.rectangle(image_display, (left, top), (right, bottom), (59, 130, 246), 3)

    # Label each face with a number
    cv2.putText(image_display, f"Face {i+1}", (left, top - 10),
                cv2.FONT_HERSHEY_SIMPLEX, 0.8, (59, 130, 246), 2)

# Step 5: Display the result
plt.figure(figsize=(10, 7))
plt.imshow(image_display)
plt.title(f"✅ Face Detection Result — {len(face_locations)} face(s) found", fontsize=14)
plt.axis("off")
plt.tight_layout()
plt.show()

print("\n✅ Phase 2 complete! Faces detected and highlighted.")
print("➡️  Move to Phase 3: Loading known faces for recognition.")
```

```python
# ── PHASE 4: Face Recognition ────────────────────────────────────────────────
# Goal: Upload a new photo and identify everyone in it using
#       the encodings built in Phase 3.
#
# ⚠️  Run Phase 3 BEFORE this — known_encodings and known_names must exist.

import cv2
import face_recognition
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Step 1: Upload the test image (photo to recognise)
print("📂 Upload a photo to identify faces in it...")
uploaded = files.upload()
test_filename = list(uploaded.keys())[0]

# Step 2: Load and convert the test image
test_image  = face_recognition.load_image_file(test_filename)
test_image_display = test_image.copy()

# Step 3: Find all face locations and encodings in the test image
print("🔍 Detecting and encoding faces in test image...")
face_locations  = face_recognition.face_locations(test_image)
face_encodings  = face_recognition.face_encodings(test_image, face_locations)

print(f"  → {len(face_locations)} face(s) found in the test image.")

# Step 4: Compare each detected face against known faces
for (top, right, bottom, left), face_encoding in zip(face_locations, face_encodings):

    # Compare this face's encoding to all known encodings
    # Returns a list of True/False — one per known person
    matches = face_recognition.compare_faces(known_encodings, face_encoding, tolerance=0.5)
    # tolerance: lower = stricter matching (0.4–0.6 is a good range)

    # Also compute distance scores — lower distance = better match
    face_distances = face_recognition.face_distance(known_encodings, face_encoding)

    name = "Unknown"        # default if no match
    color = (200, 50, 50)   # red box for unknown

    if True in matches:
        # Pick the known person with the smallest distance (best match)
        best_index = np.argmin(face_distances)
        if matches[best_index]:
            name  = known_names[best_index]
            color = (34, 197, 94)   # green box for known person

    # Step 5: Draw box and label on the image
    cv2.rectangle(test_image_display,
                  (left, top), (right, bottom), color, 3)

    # Filled rectangle behind the name for readability
    cv2.rectangle(test_image_display,
                  (left, bottom - 28), (right, bottom), color, cv2.FILLED)

    cv2.putText(test_image_display, name,
                (left + 6, bottom - 8),
                cv2.FONT_HERSHEY_SIMPLEX, 0.75, (255, 255, 255), 2)

    print(f"  → Identified: {name} "
          f"(confidence: {round((1 - min(face_distances)) * 100, 1)}%)"
          if known_encodings else "  → No known faces loaded yet.")

# Step 6: Display result
plt.figure(figsize=(10, 7))
plt.imshow(test_image_display)
plt.title("✅ Face Recognition Result", fontsize=14)
plt.axis("off")
plt.tight_layout()
plt.show()

print("\n✅ Phase 4 complete!")
print("➡️  Move to Phase 5: Attendance System.")
```

```python
# ── PHASE 5: Complete Attendance System ──────────────────────────────────────
# Goal: Recognise faces, mark attendance once per session,
#       save to attendance.csv, and let you download it.
#
# ⚠️  Run Phase 3 first to load known_encodings and known_names.

import cv2
import face_recognition
import numpy as np
import pandas as pd
import os
from datetime import datetime
from google.colab import files
import matplotlib.pyplot as plt

# ── Attendance helpers ────────────────────────────────────────────────────────

CSV_FILE = "attendance.csv"

def load_attendance():
    """Load existing attendance CSV or create a fresh one."""
    if os.path.exists(CSV_FILE):
        return pd.read_csv(CSV_FILE)
    else:
        return pd.DataFrame(columns=["Name", "Date", "Time", "Status"])

def mark_attendance(name, df, marked_today):
    """
    Add a new attendance record only if this person hasn't
    been marked yet in the current session.
    Returns updated df and marked_today set.
    """
    if name == "Unknown" or name in marked_today:
        return df, marked_today

    now  = datetime.now()
    date = now.strftime("%Y-%m-%d")
    time = now.strftime("%H:%M:%S")

    # Add new row to the dataframe
    new_row = pd.DataFrame([{
        "Name"  : name,
        "Date"  : date,
        "Time"  : time,
        "Status": "Present"
    }])
    df = pd.concat([df, new_row], ignore_index=True)
    marked_today.add(name)

    print(f"  ✅ Attendance marked → {name} | {date} | {time}")
    return df, marked_today

def save_attendance(df):
    """Save attendance dataframe to CSV."""
    df.to_csv(CSV_FILE, index=False)

# ── Main attendance function ──────────────────────────────────────────────────

def run_attendance_on_image(image_path, known_encodings, known_names):
    """
    Run the full attendance pipeline on a single image.
    Works perfectly in Google Colab (no webcam needed).
    """
    # Load current attendance records
    attendance_df = load_attendance()
    marked_today  = set()   # track who's been marked this session

    # Load and process the image
    test_image        = face_recognition.load_image_file(image_path)
    display_image     = test_image.copy()

    print(f"\n🔍 Processing image: {image_path}")

    # Detect all faces
    face_locations = face_recognition.face_locations(test_image)
    face_encodings_list = face_recognition.face_encodings(test_image, face_locations)

    print(f"  → {len(face_locations)} face(s) detected.")

    for (top, right, bottom, left), face_enc in zip(face_locations, face_encodings_list):

        name  = "Unknown"
        color = (220, 53, 69)   # red for unknown

        if known_encodings:
            matches        = face_recognition.compare_faces(known_encodings, face_enc, tolerance=0.5)
            face_distances = face_recognition.face_distance(known_encodings, face_enc)

            if True in matches:
                best_idx = np.argmin(face_distances)
                if matches[best_idx]:
                    name  = known_names[best_idx]
                    color = (40, 167, 69)   # green for known

        # Mark attendance
        attendance_df, marked_today = mark_attendance(name, attendance_df, marked_today)

        # Draw bounding box
        cv2.rectangle(display_image, (left, top), (right, bottom), color, 3)

        # Name label background
        cv2.rectangle(display_image,
                      (left, bottom - 30), (right, bottom), color, cv2.FILLED)

        # Name text
        cv2.putText(display_image, name,
                    (left + 6, bottom - 8),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)

    # Save attendance CSV
    save_attendance(attendance_df)

    # Show result image
    plt.figure(figsize=(11, 7))
    plt.imshow(display_image)
    plt.title(f"🎓 Attendance System — {len(marked_today)} person(s) marked Present", fontsize=14)
    plt.axis("off")
    plt.tight_layout()
    plt.show()

    return attendance_df

# ── Run it ────────────────────────────────────────────────────────────────────
print("📂 Upload a photo to process attendance...")
uploaded = files.upload()
test_path = list(uploaded.keys())[0]

# Run the full attendance pipeline
attendance_df = run_attendance_on_image(test_path, known_encodings, known_names)

# Show current attendance table
print("\n📋 Current Attendance Record:")
print(attendance_df.to_string(index=False))

# Download the CSV
print("\n💾 Saving and downloading attendance.csv...")
save_attendance(attendance_df)
files.download(CSV_FILE)
print("✅ Done! attendance.csv downloaded.")
```

```python
# ── PHASE 3: Load & Encode Known Faces ──────────────────────────────────────
# Goal: Upload photos of known people, convert their faces into
#       128-number "fingerprints" (encodings) for comparison later.

import os
import face_recognition
import numpy as np
from google.colab import files

# Step 1: Create the known_faces folder
os.makedirs("known_faces", exist_ok=True)
print("📁 known_faces/ folder ready.")
print("\n📂 Upload photos of people to register.")
print("   ⚠️  Important: Name each file as the person's name.")
print("   Example: Lakshy.jpg, John.png, Priya.jpg\n")

# Step 2: Upload known face images
uploaded = files.upload()

# Save uploaded files into known_faces/
for fname, data in uploaded.items():
    path = f"known_faces/{fname}"
    with open(path, "wb") as f:
        f.write(data)
    print(f"  ✅ Saved: {path}")

# Step 3: Encode each known face
known_encodings = []   # stores the 128-number fingerprint for each person
known_names     = []   # stores the matching name for each fingerprint

print("\n🧠 Encoding known faces...")

for filename in os.listdir("known_faces"):
    # Only process image files
    if not filename.lower().endswith((".jpg", ".jpeg", ".png")):
        continue

    # Extract the person's name from the filename (remove extension)
    name = os.path.splitext(filename)[0]
    filepath = f"known_faces/{filename}"

    # Load the image and convert to RGB
    image = face_recognition.load_image_file(filepath)

    # Get the face encoding (128 numbers representing this face)
    encodings = face_recognition.face_encodings(image)

    if len(encodings) == 0:
        # No face found in this photo — skip it
        print(f"  ⚠️  No face found in {filename} — skipping. Use a clearer photo.")
        continue

    # Take the first face found in the photo
    known_encodings.append(encodings[0])
    known_names.append(name)
    print(f"  ✅ Encoded: {name}")

print(f"\n🎉 Done! {len(known_names)} person(s) registered: {known_names}")
print("➡️  Move to Phase 4: Face Recognition.")
```
